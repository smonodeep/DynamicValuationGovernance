Create a Master UR table

```python
import pandas as pd
import esankhyiki

# Common PLFS parameters
base_params = {
    "indicator_code": 3,
    "frequency_code": 1,
    "year_type_code": 1,
    "gender_code": 3,
    "age_code": 1,
    "sector_code": 3,
}

all_plfs = []

# ---------------------------------------------------------
# 1. Extract state-level data: state_code 1 to 38
# ---------------------------------------------------------

for state_code in range(1, 39):

    print(f"Extracting state_code = {state_code}")

    params = {
        **base_params,
        "state_code": state_code
    }

    df_state = esankhyiki.get_data(
        "PLFS",
        params,
        format="df"
    )

    # Keep the source state identifier explicitly
    df_state["state_code"] = state_code

    all_plfs.append(df_state)


# ---------------------------------------------------------
# 2. Extract national-level data: state_code = 99
# ---------------------------------------------------------

print("Extracting national-level data: state_code = 99")

params_national = {
    **base_params,
    "state_code": 99
}

df_national = esankhyiki.get_data(
    "PLFS",
    params_national,
    format="df"
)

df_national["state_code"] = 99

all_plfs.append(df_national)


# ---------------------------------------------------------
# 3. Concatenate everything
# ---------------------------------------------------------

plfs_master = pd.concat(
    all_plfs,
    ignore_index=True
)

print("Master shape:", plfs_master.shape)
print(plfs_master.head())
print(plfs_master.tail())
plfs_master.to_csv('UR.csv', index=False)
```
